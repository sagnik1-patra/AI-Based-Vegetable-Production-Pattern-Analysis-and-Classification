```python
import os
import json
import pickle
import warnings

import numpy as np
import pandas as pd
import yaml
import h5py

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler, LabelEncoder

from sklearn.ensemble import (
    RandomForestClassifier,
    GradientBoostingClassifier,
    ExtraTreesClassifier
)

from sklearn.tree import DecisionTreeClassifier
from sklearn.neighbors import KNeighborsClassifier
from sklearn.linear_model import LogisticRegression

from sklearn.metrics import (
    accuracy_score,
    classification_report,
    confusion_matrix
)

warnings.filterwarnings("ignore")


# =============================================================================
# 2. FILE PATHS
# =============================================================================

DATASET_PATH = (
    r"C:\Users\sagni\Downloads"
    r"\AI-Based Vegetable Production Pattern Analysis and Classification"
    r"\Production_of_Important_Vegetable_crops_In_Tonnes_2019_20.csv"
)

OUTPUT_DIR = (
    r"C:\Users\sagni\Downloads"
    r"\AI-Based Vegetable Production Pattern Analysis and Classification"
)

os.makedirs(
    OUTPUT_DIR,
    exist_ok=True
)


# =============================================================================
# 3. OUTPUT FILE NAMES
# =============================================================================

H5_PATH = os.path.join(
    OUTPUT_DIR,
    "vegetable_production_model.h5"
)

PKL_PATH = os.path.join(
    OUTPUT_DIR,
    "vegetable_production_model.pkl"
)

JSON_PATH = os.path.join(
    OUTPUT_DIR,
    "vegetable_production_model.json"
)

YAML_PATH = os.path.join(
    OUTPUT_DIR,
    "vegetable_production_model.yaml"
)

SCALER_PATH = os.path.join(
    OUTPUT_DIR,
    "scaler.pkl"
)

LABEL_ENCODER_PATH = os.path.join(
    OUTPUT_DIR,
    "label_encoder.pkl"
)

PROCESSED_DATA_PATH = os.path.join(
    OUTPUT_DIR,
    "processed_vegetable_data.csv"
)


# =============================================================================
# 4. CHECK DATASET
# =============================================================================

print("=" * 80)
print("AI-BASED VEGETABLE PRODUCTION PATTERN ANALYSIS")
print("=" * 80)

print("\nChecking dataset...")

if not os.path.exists(DATASET_PATH):

    raise FileNotFoundError(
        f"\nDataset not found:\n{DATASET_PATH}"
    )

print("Dataset found successfully.")


# =============================================================================
# 5. LOAD DATASET
# =============================================================================

print("\n" + "=" * 80)
print("LOADING DATASET")
print("=" * 80)

df = pd.read_csv(
    DATASET_PATH
)

print("\nDataset loaded successfully.")

print("\nDataset Shape:")
print(df.shape)

print("\nColumns:")
for i, column in enumerate(df.columns, 1):
    print(f"{i}. {column}")

print("\nFirst 5 rows:")
print(df.head())


# =============================================================================
# 6. CLEAN COLUMN NAMES
# =============================================================================

print("\n" + "=" * 80)
print("CLEANING COLUMN NAMES")
print("=" * 80)

df.columns = (
    df.columns
    .astype(str)
    .str.strip()
    .str.replace("\n", " ", regex=False)
    .str.replace(r"\s+", " ", regex=True)
)

print("\nCleaned columns:")

for column in df.columns:
    print(column)


# =============================================================================
# 7. REMOVE COMPLETELY EMPTY ROWS
# =============================================================================

original_rows = len(df)

df = df.dropna(
    how="all"
).copy()

print(
    "\nCompletely empty rows removed:",
    original_rows - len(df)
)


# =============================================================================
# 8. IDENTIFY TALUK COLUMN
# =============================================================================

taluk_column = None

for column in df.columns:

    name = column.lower()

    if (
        "taluk" in name
        or "district" in name
        or "area" in name
        or "region" in name
    ):

        taluk_column = column
        break


if taluk_column is None:

    # Dataset's first column is treated as location column
    taluk_column = df.columns[0]


print(
    "\nLocation/Taluk column:",
    taluk_column
)


# =============================================================================
# 9. CLEAN TALUK VALUES
# =============================================================================

df[taluk_column] = (
    df[taluk_column]
    .astype(str)
    .str.strip()
)


# =============================================================================
# 10. REMOVE AGGREGATE/TOTAL ROWS
# =============================================================================

print("\n" + "=" * 80)
print("REMOVING AGGREGATE ROWS")
print("=" * 80)

aggregate_pattern = (
    r"state\s*total|grand\s*total|total$|all\s*india"
)

aggregate_mask = (
    df[taluk_column]
    .astype(str)
    .str.lower()
    .str.contains(
        aggregate_pattern,
        regex=True,
        na=False
    )
)

print(
    "\nAggregate rows found:",
    aggregate_mask.sum()
)

if aggregate_mask.sum() > 0:

    print("\nAggregate rows:")

    print(
        df.loc[
            aggregate_mask,
            [taluk_column]
        ]
    )


df = df[
    ~aggregate_mask
].copy()

df.reset_index(
    drop=True,
    inplace=True
)

print(
    "\nRows remaining:",
    len(df)
)


# =============================================================================
# 11. IDENTIFY VEGETABLE PRODUCTION COLUMNS
# =============================================================================

print("\n" + "=" * 80)
print("IDENTIFYING PRODUCTION COLUMNS")
print("=" * 80)

production_columns = []

for column in df.columns:

    if column == taluk_column:
        continue

    # Attempt conversion
    converted = pd.to_numeric(
        df[column],
        errors="coerce"
    )

    # If a meaningful amount of the column is numeric,
    # treat it as a production feature.
    if converted.notna().sum() > 0:

        df[column] = converted

        production_columns.append(
            column
        )


if len(production_columns) == 0:

    raise ValueError(
        "No numeric vegetable production columns were found."
    )


print(
    "\nProduction columns found:",
    len(production_columns)
)

for column in production_columns:
    print("-", column)


# =============================================================================
# 12. HANDLE MISSING VALUES
# =============================================================================

print("\n" + "=" * 80)
print("HANDLING MISSING VALUES")
print("=" * 80)

print("\nMissing values before preprocessing:")

print(
    df[
        production_columns
    ].isnull().sum()
)


# -----------------------------------------------------------------------------
# Rows where ALL crop production values are missing
# -----------------------------------------------------------------------------

all_missing_mask = (
    df[production_columns]
    .isnull()
    .all(axis=1)
)

print(
    "\nRows with all production values missing:",
    all_missing_mask.sum()
)


if all_missing_mask.sum() > 0:

    print("\nTaluks with all production values missing:")

    print(
        df.loc[
            all_missing_mask,
            taluk_column
        ].tolist()
    )


# Remove completely uninformative production rows
df = df[
    ~all_missing_mask
].copy()


# -----------------------------------------------------------------------------
# Fill remaining individual missing values using median
# -----------------------------------------------------------------------------

for column in production_columns:

    median_value = df[
        column
    ].median()

    df[column] = (
        df[column]
        .fillna(
            median_value
        )
    )


df.reset_index(
    drop=True,
    inplace=True
)


print("\nMissing values after preprocessing:")

print(
    df[
        production_columns
    ].isnull().sum()
)


# =============================================================================
# 13. HANDLE NEGATIVE VALUES
# =============================================================================

# Production cannot logically be negative.

for column in production_columns:

    df[column] = df[column].clip(
        lower=0
    )


# =============================================================================
# 14. CREATE TOTAL PRODUCTION
# =============================================================================

print("\n" + "=" * 80)
print("CREATING TOTAL VEGETABLE PRODUCTION")
print("=" * 80)

df["Total_Production"] = (
    df[
        production_columns
    ]
    .sum(axis=1)
)


print("\nTotal Production Statistics:")

print(
    df["Total_Production"]
    .describe()
)


# =============================================================================
# 15. IDENTIFY DOMINANT VEGETABLE
# =============================================================================

df["Dominant_Vegetable"] = (
    df[
        production_columns
    ]
    .idxmax(axis=1)
)


print("\nDominant vegetable distribution:")

print(
    df[
        "Dominant_Vegetable"
    ].value_counts()
)


# =============================================================================
# 16. CREATE PRODUCTION CATEGORY
# =============================================================================

print("\n" + "=" * 80)
print("CREATING PRODUCTION CATEGORIES")
print("=" * 80)

# We create Low / Medium / High classes based on production quantiles.
#
# Rank is used first to avoid qcut problems when multiple taluks
# contain exactly the same production value.

production_rank = (
    df["Total_Production"]
    .rank(
        method="first"
    )
)


df["Production_Category"] = pd.qcut(
    production_rank,
    q=3,
    labels=[
        "Low",
        "Medium",
        "High"
    ]
)


df[
    "Production_Category"
] = (
    df[
        "Production_Category"
    ]
    .astype(str)
)


print("\nProduction Category Distribution:")

print(
    df[
        "Production_Category"
    ].value_counts()
)


# =============================================================================
# 17. DISPLAY SAMPLE PROCESSED DATA
# =============================================================================

print("\nProcessed Dataset Sample:")

display_columns = [
    taluk_column,
    "Total_Production",
    "Dominant_Vegetable",
    "Production_Category"
]

print(
    df[
        display_columns
    ].head(10)
)


# =============================================================================
# 18. SAVE PROCESSED DATASET
# =============================================================================

df.to_csv(
    PROCESSED_DATA_PATH,
    index=False
)

print(
    "\nProcessed dataset saved:"
)

print(
    PROCESSED_DATA_PATH
)


# =============================================================================
# 19. PREPARE MACHINE LEARNING FEATURES
# =============================================================================

print("\n" + "=" * 80)
print("PREPARING MACHINE LEARNING DATA")
print("=" * 80)

X = df[
    production_columns
].copy()

y = df[
    "Production_Category"
].copy()


print(
    "\nFeature matrix shape:",
    X.shape
)

print(
    "Target shape:",
    y.shape
)


# =============================================================================
# 20. ENCODE TARGET
# =============================================================================

label_encoder = LabelEncoder()

y_encoded = label_encoder.fit_transform(
    y
)


print("\nTarget Encoding:")

for class_name, encoded_value in zip(
    label_encoder.classes_,
    label_encoder.transform(
        label_encoder.classes_
    )
):

    print(
        f"{class_name} -> {encoded_value}"
    )


# =============================================================================
# 21. TRAIN TEST SPLIT
# =============================================================================

X_train, X_test, y_train, y_test = (
    train_test_split(
        X,
        y_encoded,
        test_size=0.20,
        random_state=42,
        stratify=y_encoded
    )
)


print(
    "\nTraining samples:",
    len(X_train)
)

print(
    "Testing samples:",
    len(X_test)
)


# =============================================================================
# 22. FEATURE SCALING
# =============================================================================

scaler = StandardScaler()

X_train_scaled = scaler.fit_transform(
    X_train
)

X_test_scaled = scaler.transform(
    X_test
)


print(
    "\nFeature scaling completed."
)


# =============================================================================
# 23. CREATE MACHINE LEARNING MODELS
# =============================================================================

print("\n" + "=" * 80)
print("CREATING MACHINE LEARNING MODELS")
print("=" * 80)

models = {

    "Random Forest":
        RandomForestClassifier(
            n_estimators=300,
            random_state=42,
            class_weight="balanced"
        ),

    "Extra Trees":
        ExtraTreesClassifier(
            n_estimators=300,
            random_state=42,
            class_weight="balanced"
        ),

    "Gradient Boosting":
        GradientBoostingClassifier(
            n_estimators=200,
            learning_rate=0.05,
            random_state=42
        ),

    "Decision Tree":
        DecisionTreeClassifier(
            random_state=42,
            class_weight="balanced"
        ),

    "KNN":
        KNeighborsClassifier(
            n_neighbors=5
        ),

    "Logistic Regression":
        LogisticRegression(
            max_iter=5000,
            class_weight="balanced",
            random_state=42
        )
}


# =============================================================================
# 24. TRAIN AND EVALUATE MODELS
# =============================================================================

print("\n" + "=" * 80)
print("TRAINING MODELS")
print("=" * 80)

model_results = {}

trained_models = {}


for model_name, model in models.items():

    print("\n" + "-" * 80)

    print(
        "Training:",
        model_name
    )

    print("-" * 80)


    # -------------------------------------------------------------------------
    # Models that benefit from scaled features
    # -------------------------------------------------------------------------

    if model_name in [
        "KNN",
        "Logistic Regression"
    ]:

        model.fit(
            X_train_scaled,
            y_train
        )

        predictions = model.predict(
            X_test_scaled
        )

    else:

        model.fit(
            X_train,
            y_train
        )

        predictions = model.predict(
            X_test
        )


    accuracy = accuracy_score(
        y_test,
        predictions
    )


    model_results[
        model_name
    ] = float(accuracy)

    trained_models[
        model_name
    ] = model


    print(
        f"\nAccuracy: {accuracy:.4f}"
    )

    print(
        f"Accuracy (%): {accuracy * 100:.2f}%"
    )


# =============================================================================
# 25. DISPLAY MODEL COMPARISON
# =============================================================================

print("\n" + "=" * 80)
print("MODEL COMPARISON")
print("=" * 80)

sorted_results = sorted(
    model_results.items(),
    key=lambda item: item[1],
    reverse=True
)


for rank, (
    model_name,
    accuracy
) in enumerate(
    sorted_results,
    1
):

    print(
        f"{rank}. "
        f"{model_name}: "
        f"{accuracy * 100:.2f}%"
    )


# =============================================================================
# 26. SELECT BEST MODEL
# =============================================================================

best_model_name = max(
    model_results,
    key=model_results.get
)

best_accuracy = model_results[
    best_model_name
]

best_model = trained_models[
    best_model_name
]


print("\n" + "=" * 80)
print("BEST MODEL")
print("=" * 80)

print(
    "\nBest Model:",
    best_model_name
)

print(
    f"Best Accuracy: "
    f"{best_accuracy * 100:.2f}%"
)


# =============================================================================
# 27. FINAL BEST MODEL PREDICTIONS
# =============================================================================

if best_model_name in [
    "KNN",
    "Logistic Regression"
]:

    best_predictions = (
        best_model.predict(
            X_test_scaled
        )
    )

else:

    best_predictions = (
        best_model.predict(
            X_test
        )
    )


# =============================================================================
# 28. CLASSIFICATION REPORT
# =============================================================================

print("\n" + "=" * 80)
print("CLASSIFICATION REPORT")
print("=" * 80)

report = classification_report(
    y_test,
    best_predictions,
    target_names=label_encoder.classes_,
    zero_division=0
)

print(
    "\n",
    report
)


# =============================================================================
# 29. CONFUSION MATRIX
# =============================================================================

cm = confusion_matrix(
    y_test,
    best_predictions
)

print("\nConfusion Matrix:")

print(cm)


# =============================================================================
# 30. TRAIN BEST MODEL ON COMPLETE DATASET
# =============================================================================

print("\n" + "=" * 80)
print("TRAINING FINAL MODEL ON COMPLETE DATASET")
print("=" * 80)

# Fit a final scaler using complete dataset.

final_scaler = StandardScaler()

X_scaled_full = (
    final_scaler.fit_transform(
        X
    )
)


if best_model_name in [
    "KNN",
    "Logistic Regression"
]:

    best_model.fit(
        X_scaled_full,
        y_encoded
    )

else:

    best_model.fit(
        X,
        y_encoded
    )


print(
    "\nFinal model trained successfully."
)


# =============================================================================
# 31. SAVE BEST MODEL AS PKL
# =============================================================================

print("\n" + "=" * 80)
print("SAVING PKL FILE")
print("=" * 80)

model_package = {

    "model":
        best_model,

    "scaler":
        final_scaler,

    "label_encoder":
        label_encoder,

    "feature_columns":
        production_columns,

    "model_name":
        best_model_name,

    "accuracy":
        best_accuracy,

    "requires_scaling":
        best_model_name in [
            "KNN",
            "Logistic Regression"
        ]
}


with open(
    PKL_PATH,
    "wb"
) as file:

    pickle.dump(
        model_package,
        file
    )


print(
    "\nPKL saved successfully:"
)

print(
    PKL_PATH
)


# =============================================================================
# 32. SAVE SCALER
# =============================================================================

with open(
    SCALER_PATH,
    "wb"
) as file:

    pickle.dump(
        final_scaler,
        file
    )


print(
    "\nScaler saved:"
)

print(
    SCALER_PATH
)


# =============================================================================
# 33. SAVE LABEL ENCODER
# =============================================================================

with open(
    LABEL_ENCODER_PATH,
    "wb"
) as file:

    pickle.dump(
        label_encoder,
        file
    )


print(
    "\nLabel encoder saved:"
)

print(
    LABEL_ENCODER_PATH
)


# =============================================================================
# 34. SAVE MODEL INFORMATION AS JSON
# =============================================================================

print("\n" + "=" * 80)
print("SAVING JSON FILE")
print("=" * 80)

json_data = {

    "project":
        "AI-Based Vegetable Production Pattern Analysis and Classification",

    "dataset":
        os.path.basename(
            DATASET_PATH
        ),

    "best_model":
        best_model_name,

    "best_accuracy":
        float(
            best_accuracy
        ),

    "best_accuracy_percentage":
        float(
            best_accuracy * 100
        ),

    "number_of_records":
        int(
            len(df)
        ),

    "number_of_features":
        int(
            len(production_columns)
        ),

    "features":
        [
            str(column)
            for column
            in production_columns
        ],

    "target":
        "Production_Category",

    "classes":
        [
            str(value)
            for value
            in label_encoder.classes_
        ],

    "class_mapping": {

        str(class_name):
            int(encoded_value)

        for class_name, encoded_value
        in zip(
            label_encoder.classes_,
            label_encoder.transform(
                label_encoder.classes_
            )
        )
    },

    "model_comparison": {

        model:
            float(accuracy)

        for model, accuracy
        in model_results.items()
    },

    "preprocessing": {

        "missing_values":
            "Median imputation",

        "aggregate_rows":
            "Removed",

        "scaling":
            "StandardScaler",

        "classification_method":
            "Quantile-based Low/Medium/High production classes"
    }
}


with open(
    JSON_PATH,
    "w",
    encoding="utf-8"
) as file:

    json.dump(
        json_data,
        file,
        indent=4
    )


print(
    "\nJSON saved successfully:"
)

print(
    JSON_PATH
)


# =============================================================================
# 35. SAVE MODEL INFORMATION AS YAML
# =============================================================================

print("\n" + "=" * 80)
print("SAVING YAML FILE")
print("=" * 80)

yaml_data = {

    "project":
        "AI-Based Vegetable Production Pattern Analysis and Classification",

    "dataset":
        os.path.basename(
            DATASET_PATH
        ),

    "best_model":
        str(
            best_model_name
        ),

    "best_accuracy":
        float(
            best_accuracy
        ),

    "best_accuracy_percentage":
        float(
            best_accuracy * 100
        ),

    "records":
        int(
            len(df)
        ),

    "features":
        [
            str(column)
            for column
            in production_columns
        ],

    "target":
        "Production_Category",

    "classes":
        [
            str(value)
            for value
            in label_encoder.classes_
        ],

    "model_results": {

        str(model):
            float(accuracy)

        for model, accuracy
        in model_results.items()
    }
}


with open(
    YAML_PATH,
    "w",
    encoding="utf-8"
) as file:

    yaml.safe_dump(
        yaml_data,
        file,
        sort_keys=False,
        allow_unicode=True
    )


print(
    "\nYAML saved successfully:"
)

print(
    YAML_PATH
)


# =============================================================================
# 36. SAVE MODEL INFORMATION AS H5
# =============================================================================

print("\n" + "=" * 80)
print("SAVING H5 FILE")
print("=" * 80)

with h5py.File(
    H5_PATH,
    "w"
) as h5_file:

    # -------------------------------------------------------------------------
    # Metadata
    # -------------------------------------------------------------------------

    h5_file.attrs[
        "project"
    ] = (
        "AI-Based Vegetable Production Pattern Analysis "
        "and Classification"
    )

    h5_file.attrs[
        "best_model"
    ] = best_model_name

    h5_file.attrs[
        "accuracy"
    ] = best_accuracy

    h5_file.attrs[
        "number_of_records"
    ] = len(df)

    h5_file.attrs[
        "number_of_features"
    ] = len(
        production_columns
    )


    # -------------------------------------------------------------------------
    # Feature names
    # -------------------------------------------------------------------------

    string_dtype = h5py.string_dtype(
        encoding="utf-8"
    )

    h5_file.create_dataset(
        "feature_names",
        data=np.array(
            production_columns,
            dtype=object
        ),
        dtype=string_dtype
    )


    # -------------------------------------------------------------------------
    # Classes
    # -------------------------------------------------------------------------

    h5_file.create_dataset(
        "classes",
        data=np.array(
            label_encoder.classes_,
            dtype=object
        ),
        dtype=string_dtype
    )


    # -------------------------------------------------------------------------
    # Dataset
    # -------------------------------------------------------------------------

    h5_file.create_dataset(
        "X",
        data=X.to_numpy(
            dtype=np.float64
        )
    )

    h5_file.create_dataset(
        "y",
        data=np.asarray(
            y_encoded,
            dtype=np.int32
        )
    )


    # -------------------------------------------------------------------------
    # Scaler information
    # -------------------------------------------------------------------------

    scaler_group = (
        h5_file.create_group(
            "scaler"
        )
    )

    scaler_group.create_dataset(
        "mean",
        data=final_scaler.mean_
    )

    scaler_group.create_dataset(
        "scale",
        data=final_scaler.scale_
    )

    scaler_group.create_dataset(
        "var",
        data=final_scaler.var_
    )


    # -------------------------------------------------------------------------
    # Model comparison
    # -------------------------------------------------------------------------

    comparison_group = (
        h5_file.create_group(
            "model_comparison"
        )
    )

    for model_name, accuracy in model_results.items():

        safe_name = (
            model_name
            .replace(
                " ",
                "_"
            )
            .lower()
        )

        comparison_group.attrs[
            safe_name
        ] = accuracy


    # -------------------------------------------------------------------------
    # Store serialized model inside H5
    # -------------------------------------------------------------------------

    serialized_model = pickle.dumps(
        best_model
    )

    serialized_array = np.frombuffer(
        serialized_model,
        dtype=np.uint8
    )

    h5_file.create_dataset(
        "serialized_model",
        data=serialized_array
    )


print(
    "\nH5 saved successfully:"
)

print(
    H5_PATH
)


# =============================================================================
# 37. MAKE PREDICTIONS FOR COMPLETE DATASET
# =============================================================================

print("\n" + "=" * 80)
print("GENERATING COMPLETE DATASET PREDICTIONS")
print("=" * 80)


if best_model_name in [
    "KNN",
    "Logistic Regression"
]:

    full_predictions_encoded = (
        best_model.predict(
            X_scaled_full
        )
    )

else:

    full_predictions_encoded = (
        best_model.predict(
            X
        )
    )


full_predictions = (
    label_encoder.inverse_transform(
        full_predictions_encoded
    )
)


df[
    "Predicted_Category"
] = full_predictions


# =============================================================================
# 38. SAVE FINAL DATASET WITH PREDICTIONS
# =============================================================================

FINAL_DATA_PATH = os.path.join(
    OUTPUT_DIR,
    "vegetable_production_predictions.csv"
)


df.to_csv(
    FINAL_DATA_PATH,
    index=False
)


print(
    "\nPredictions saved:"
)

print(
    FINAL_DATA_PATH
)


# =============================================================================
# 39. TEST LOADING PKL MODEL
# =============================================================================

print("\n" + "=" * 80)
print("TESTING SAVED PKL MODEL")
print("=" * 80)


with open(
    PKL_PATH,
    "rb"
) as file:

    loaded_package = pickle.load(
        file
    )


loaded_model = loaded_package[
    "model"
]

loaded_scaler = loaded_package[
    "scaler"
]

loaded_encoder = loaded_package[
    "label_encoder"
]


print(
    "\nPKL model loaded successfully."
)

print(
    "Loaded Model:",
    loaded_package[
        "model_name"
    ]
)


# =============================================================================
# 40. TEST H5 FILE
# =============================================================================

print("\n" + "=" * 80)
print("TESTING H5 FILE")
print("=" * 80)


with h5py.File(
    H5_PATH,
    "r"
) as h5_file:

    print(
        "\nH5 Groups/Datasets:"
    )

    for key in h5_file.keys():

        print(
            "-",
            key
        )

    print(
        "\nStored model:",
        h5_file.attrs[
            "best_model"
        ]
    )

    print(
        "Stored accuracy:",
        h5_file.attrs[
            "accuracy"
        ]
    )


# =============================================================================
# 41. DISPLAY FINAL RESULTS
# =============================================================================

print("\n" + "=" * 80)
print("FINAL RESULTS")
print("=" * 80)

print(
    "\nProject:"
)

print(
    "AI-Based Vegetable Production Pattern "
    "Analysis and Classification"
)

print(
    "\nBest Model:",
    best_model_name
)

print(
    f"Accuracy: "
    f"{best_accuracy * 100:.2f}%"
)

print(
    "\nNumber of Taluks Used:",
    len(df)
)

print(
    "\nNumber of Production Features:",
    len(production_columns)
)

print(
    "\nProduction Categories:"
)

for category in label_encoder.classes_:

    print(
        "-",
        category
    )


# =============================================================================
# 42. OUTPUT FILE SUMMARY
# =============================================================================

print("\n" + "=" * 80)
print("GENERATED FILES")
print("=" * 80)

generated_files = [

    H5_PATH,
    PKL_PATH,
    JSON_PATH,
    YAML_PATH,
    SCALER_PATH,
    LABEL_ENCODER_PATH,
    PROCESSED_DATA_PATH,
    FINAL_DATA_PATH
]


for file_path in generated_files:

    if os.path.exists(
        file_path
    ):

        size_kb = (
            os.path.getsize(
                file_path
            )
            / 1024
        )

        print(
            f"\n[OK] "
            f"{os.path.basename(file_path)} "
            f"({size_kb:.2f} KB)"
        )

    else:

        print(
            f"\n[ERROR] "
            f"{os.path.basename(file_path)}"
        )


print("\n" + "=" * 80)
print("PROJECT EXECUTION COMPLETED SUCCESSFULLY")
print("=" * 80)

print(
    "\nAll files have been saved in:"
)

print(
    OUTPUT_DIR
)
```
